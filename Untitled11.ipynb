```python
from collections import Counter
import pandas as pd

# Sample customer movement data: sequences of zones visited
customer_paths = [
    ['A', 'B', 'C', 'D'],
    ['A', 'B', 'E', 'D'],
    ['B', 'C', 'D'],
    ['A', 'E', 'D'],
    ['B', 'E', 'D'],
    ['A', 'B', 'C', 'E', 'D']
]

# Extract all unique zones
all_zones = sorted(set(zone for path in customer_paths for zone in path))

# Count transitions between zones
transitions = {zone: Counter() for zone in all_zones}
for path in customer_paths:
    for i in range(len(path) - 1):
        current_zone = path[i]
        next_zone = path[i + 1]
        transitions[current_zone][next_zone] += 1

# Build transition probability matrix
transition_matrix = pd.DataFrame(index=all_zones, columns=all_zones).fillna(0)
for current_zone, next_counts in transitions.items():
    total = sum(next_counts.values())
    for next_zone, count in next_counts.items():
        transition_matrix.loc[current_zone, next_zone] = count / total

print("Transition Probability Matrix:")
print(transition_matrix)

# Function to predict next zone based on current zone
def predict_next_zone(current_zone, matrix):
    if current_zone not in matrix.index:
        return None
    probs = matrix.loc[current_zone]
    if probs.sum() == 0:
        return None
    return probs.idxmax()

# Example prediction
current = 'B'
next_zone = predict_next_zone(current, transition_matrix)
print(f"\nPredicted next zone after '{current}': {next_zone}")

# Calculate total incoming transitions per zone to find high-traffic zones
incoming_counts = Counter()
for next_counts in transitions.values():
    for zone, count in next_counts.items():
        incoming_counts[zone] += count

high_traffic_zones = sorted(incoming_counts.items(), key=lambda x: x[1], reverse=True)

print("\nHigh-traffic zones (for product placement):")
for zone, count in high_traffic_zones:
    print(f"Zone {zone}: {count} visits")
```
