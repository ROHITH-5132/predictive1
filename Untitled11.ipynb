```python
from collections import Counter
import pandas as pd

# Sample customer movement data: sequences of zones visited
customer_paths = [
    ['A', 'B', 'C', 'D'],
    ['A', 'B', 'E', 'D'],
    ['B', 'C', 'D'],
    ['A', 'E', 'D'],
    ['B', 'E', 'D'],
    ['A', 'B', 'C', 'E', 'D']
]

# Extract all unique zones
all_zones = sorted(set(zone for path in customer_paths for zone in path))

# Count transitions between zones
transitions = {zone: Counter() for zone in all_zones}
for path in customer_paths:
    for i in range(len(path) - 1):
        current_zone = path[i]
        next_zone = path[i + 1]
        transitions[current_zone][next_zone] += 1

# Build transition probability matrix
transition_matrix = pd.DataFrame(index=all_zones, columns=all_zones).fillna(0)
for current_zone, next_counts in transitions.items():
    total = sum(next_counts.values())
    for next_zone, count in next_counts.items():
        transition_matrix.loc[current_zone, next_zone] = count / total

print("Transition Probability Matrix:")
print(transition_matrix)

# Function to predict next zone based on current zone
def predict_next_zone(current_zone, matrix):
    if current_zone not in matrix.index:
        return None
    probs = matrix.loc[current_zone]
    if probs.sum() == 0:
        return None
    return probs.idxmax()

# Example prediction
current = 'B'
next_zone = predict_next_zone(current, transition_matrix)
print(f"\nPredicted next zone after '{current}': {next_zone}")

# Calculate total incoming transitions per zone to find high-traffic zones
incoming_counts = Counter()
for next_counts in transitions.values():
    for zone, count in next_counts.items():
        incoming_counts[zone] += count

high_traffic_zones = sorted(incoming_counts.items(), key=lambda x: x[1], reverse=True)

print("\nHigh-traffic zones (for product placement):")
for zone, count in high_traffic_zones:
    print(f"Zone {zone}: {count} visits")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Sample dummy training data
# Each row: [price, competitor_price, clicks]
X = np.array([
    [60, 58, 100],
    [65, 60, 120],
    [70, 64, 130],
    [75, 67, 115],
    [80, 70, 110],
    [85, 73, 105],
    [90, 76, 95],
    [95, 80, 90],
    [100, 82, 80],
    [105, 85, 75],
])

# Purchases (demand) for each row
y = np.array([40, 42, 50, 47, 43, 40, 37, 30, 25, 20])  # Target variable

# Step 2: Train the Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# Step 3: Predict demand and revenue for price range
price_range = np.arange(50, 151, 1)
avg_competitor_price = np.mean(X[:, 1])
avg_clicks = np.mean(X[:, 2])

revenues = []
demands = []

for p in price_range:
    # Predict demand using trained model
    features = np.array([[p, avg_competitor_price, avg_clicks]])
    predicted_demand = model.predict(features)[0]
    predicted_demand = max(predicted_demand, 0)  # No negative demand

    revenue = p * predicted_demand
    demands.append(predicted_demand)
    revenues.append(revenue)

# Step 4: Find optimal price
optimal_index = np.argmax(revenues)
optimal_price = price_range[optimal_index]
optimal_revenue = revenues[optimal_index]
optimal_demand = demands[optimal_index]

# Step 5: Print results
print(f"✅ Optimal Price: ₹{optimal_price}")
print(f"📦 Expected Demand: {optimal_demand:.2f} units")
print(f"💰 Expected Revenue: ₹{optimal_revenue:.2f}")

# Step 6: Plot Revenue Curve
plt.figure(figsize=(10, 6))
plt.plot(price_range, revenues, label='Revenue', color='blue')
plt.axvline(optimal_price, color='red', linestyle='--', label=f'Optimal Price ₹{optimal_price}')
plt.xlabel('Price')
plt.ylabel('Revenue')
plt.title('Price vs Revenue Curve')
plt.legend()
plt.grid(True)
plt.show()
```
